```python
categorical_cols = [
    'Contract',
    'InternetService',
    'PaymentMethod',
    'TechSupport',
    'OnlineSecurity'
]

for col in categorical_cols:
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x=col, hue='Churn')
    plt.title(f'{col} vs Churn')
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()
```

```python
# Churn analysis for important categorical features

features = [
    'Contract',
    'PaymentMethod',
    'InternetService',
    'OnlineSecurity',
    'TechSupport',
    'PaperlessBilling'
]

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for ax, feature in zip(axes.flatten(), features):
    sns.countplot(data=df, x=feature, hue='Churn', ax=ax)
    ax.set_title(f'{feature} vs Churn')
    ax.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

```python
# Numerical features vs Churn

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(data=df, x='Churn', y='tenure', ax=axes[0])
axes[0].set_title('Tenure vs Churn')

sns.boxplot(data=df, x='Churn', y='MonthlyCharges', ax=axes[1])
axes[1].set_title('Monthly Charges vs Churn')

sns.boxplot(data=df, x='Churn', y='TotalCharges', ax=axes[2])
axes[2].set_title('Total Charges vs Churn')

plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap for numerical variables

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10, 7))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
# Correlation analysis for numerical features

# Convert TotalCharges to numeric
# Invalid/blank values will become NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Convert Churn from Yes/No to 1/0
df['Churn_numeric'] = df['Churn'].map({'Yes': 1, 'No': 0})

# Select numerical columns
numeric_cols = [
    'SeniorCitizen',
    'tenure',
    'MonthlyCharges',
    'TotalCharges',
    'Churn_numeric'
]

# Calculate correlation
correlation = df[numeric_cols].corr()

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

```python
# Clean TotalCharges
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Convert Churn from Yes/No to 1/0
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# Check missing values created during conversion
print("Missing values:")
print(df[['TotalCharges', 'Churn']].isnull().sum())

# Fill missing TotalCharges with median
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())

# Select numerical columns
numeric_cols = ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']

# Calculate correlation
correlation = df[numeric_cols].corr()

# Display correlation matrix
print(correlation)

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```
